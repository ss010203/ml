```python
import pandas as pd

url = "https://raw.githubusercontent.com/nguyen-toan/ISLR/master/dataset/Advertising.csv"
df = pd.read_csv(url)
df.head()
```

```python
print("Missing Values: ", df.isna().sum())
print("Duplicate Values: ", df.duplicated().sum())
```

```python
df = df.drop(columns=['Unnamed: 0'])
df.drop_duplicates()
df.dropna()
df.head()
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns=['Sales'])
y = df['Sales']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
predicitons = model.predict(x_test)
```

```python
df_results = pd.DataFrame({'Actual': y_test, 'Predicted': predicitons, 'Residual': abs(y_test - predicitons)})
df_results = df_results.sort_values(['Residual'], ascending=False)
df_results.head(5).round(2)
```

```python
import matplotlib.pyplot as plt

plt.scatter(y_test, predicitons)
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.title("Actual vs Predicted")
plt.show()
```
