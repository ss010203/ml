```python
import pandas as pd

url = "https://huggingface.co/api/resolve-cache/datasets/Rodrigopiva/Social_Network_Ads.csv/467e4443107c31a94450f272ca14d73e6af915db/Social_Network_Ads.csv?utm_source=chatgpt.com&etag=%224a53849c2baf5c3c6af4e55c3f727734e4ff21e6%22"
df = pd.read_csv(url)
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

x = df[["Age", "EstimatedSalary"]]
y = df["Purchased"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify = y)
model = SVC(kernel="rbf")
model.fit(x_train, y_train)
pred = model.predict(x_test)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

model_scaled = SVC(kernel="rbf")
model_scaled.fit(x_train_scaled, y_train)
pred_scaled = model_scaled.predict(x_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score
print("RBF Kernel SVM using unscaled features:")
print("Confusion Matrix:")
print(confusion_matrix(y_test, pred))
print("Accuracy: ", accuracy_score(y_test, pred))
print("F1 Score: ", f1_score(y_test, pred))

print("\nRBF Kernel SVM using scaled features:")
print("Confusion Matrix:")
print(confusion_matrix(y_test, pred_scaled))
print("Accuracy: ", accuracy_score(y_test, pred_scaled))
print("F1 Score: ", f1_score(y_test, pred_scaled))
```
