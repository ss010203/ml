```python
import pandas as pd

url = "https://huggingface.co/api/resolve-cache/datasets/Rodrigopiva/Social_Network_Ads.csv/467e4443107c31a94450f272ca14d73e6af915db/Social_Network_Ads.csv?utm_source=chatgpt.com&etag=%224a53849c2baf5c3c6af4e55c3f727734e4ff21e6%22"
df = pd.read_csv(url)
df.head()
```

```python
print("Missing Vals: ", df.isnull().sum().sum())
print("Classes: ", df["Purchased"].value_counts().to_dict())
```

```python
from sklearn.model_selection import train_test_split

x = df[["Age", "EstimatedSalary"]]
y = df["Purchased"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

for kernel in ["linear", "rbf"]:
  pred = SVC(kernel=kernel).fit(x_train, y_train).predict(x_test)
  print(f"Kernel: {kernel.upper()}")
  print(f"SVM Confusion Matrix [[TN FP] [FN TP]]: {confusion_matrix(y_test, pred).tolist()}")
  print(f"SVM Accuracy: {accuracy_score(y_test, pred)}")
  print(f"SVM Precision: {precision_score(y_test, pred)}")
  print(f"SVM Recall: {recall_score(y_test, pred)}")
  print(f"SVM F1 Score: {f1_score(y_test, pred)}\n")
```
