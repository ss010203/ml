```python
#loading dataset, dropping irrelevant columns & removing duplicate & missing vals (all done in first task)
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/nguyen-toan/ISLR/master/dataset/Advertising.csv"
df = pd.read_csv(url)

df = df.drop(columns=['Unnamed: 0'])
df = df.drop_duplicates()
df = df.dropna()
df.head()
```

```python
#test train split
from sklearn.model_selection import train_test_split
features = ['TV', 'Radio']

x_train,x_test,   y_train,y_test, = train_test_split(df[features], df["Sales"], test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

model = LinearRegression()
model.fit(x_train[features], y_train)

pred = model.predict(x_test)
print("RMSE: ", np.sqrt(mean_squared_error(y_test, pred)))
print("R2: ", r2_score(y_test, pred))
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2, include_bias=False)
x_train_poly = poly.fit_transform(x_train)

model_2 = LinearRegression()
model_2.fit(x_train_poly, y_train)

pred_2 = model_2.predict(poly.fit_transform(x_test))
print("RMSE: ", np.sqrt(mean_squared_error(y_test, pred_2)))
print("R2: ", r2_score(y_test, pred_2))
```

```python
print("Linear: RMSE=", np.sqrt(mean_squared_error(y_test, pred)), "\tR2=", r2_score(y_test, pred))
print("Poly: RMSE=", np.sqrt(mean_squared_error(y_test, pred_2)), "\tR2=", r2_score(y_test, pred_2))
```
